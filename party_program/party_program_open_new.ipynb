```python

party_name = "Programm_Linke_MV"
file_type = ".pdf"

pdf_name = party_name + file_type # Create a name for the PDF file
txt_name = pdf_name.replace(".pdf", ".txt") # Create a name for the text file


import fitz # PyMuPDF
# Open the PDF file
doc = fitz.open(pdf_name)
# Iterate through pages and extract text

all_text = ""
for page_num in range(doc.page_count):
   page = doc.load_page(page_num) # Load a specific page
   text = page.get_text() # Extract text from the page
   #print(f"Page {page_num + 1}:\n{text}")
   all_text += text + "\n" # Append the text of the page to the overall text

doc.close()




# Python program to demonstrate
# saving a text file

file = open(txt_name, 'w')
file.write(all_text)
file.close()
```
