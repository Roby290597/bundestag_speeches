### Vektorisierung des Parteiprogramms

```python
## Load the party program data 
print("Loading party program data...")
print(os.listdir())
```

```python
client = chromadb.Client()
collection = client.create_collection("party_programs")
```

```python
with open("AfD_Parteiprogramm2026.txt", "r", encoding="utf-8") as f:
    sections: list[str] = f.read().splitlines()
```

```python
collection.add(
    ids=[str(uuid.uuid4()) for _ in range(len(sections))],
    documents=sections,
    metadatas=[{"source": "AfD_Parteiprogramm2026.txt"} for _ in range(len(sections))],
)
```

```python
results = collection.query(
    query_texts=["Was ist die Stellungnahme der Partei zum Klimaschutz?"],
    n_results=5
)
```

```python
for result in results['documents']:
    print(result)
```
